# Teste t por senioridade

Comparação das médias salariais de homens e mulheres dentro de cada nível de senioridade.

```python
import pandas as pd
from scipy import stats

df = pd.read_csv("../data/synthetic/fato_mercado_salarios_tech.csv")
```

```python
senioridades = [
    "Junior",
    "Pleno",
    "Senior",
    "Lead / Staff Specialist",
    "Diretoria / C-Level"
]

resultados = []

for senioridade in senioridades:
    grupo = df[df["Senioridade"] == senioridade]

    homens = grupo.loc[
        grupo["Genero"] == "Masculino",
        "Salario_Mensal_BRL"
    ]

    mulheres = grupo.loc[
        grupo["Genero"] == "Feminino",
        "Salario_Mensal_BRL"
    ]

    t_stat, p_val = stats.ttest_ind(
        homens,
        mulheres,
        equal_var=False
    )

    media_homens = homens.mean()
    media_mulheres = mulheres.mean()
    gap_pct = (
        (media_homens - media_mulheres)
        / media_homens
    ) * 100

    resultados.append({
        "Senioridade": senioridade,
        "N_Homens": len(homens),
        "N_Mulheres": len(mulheres),
        "Media_Homens_R$": media_homens,
        "Media_Mulheres_R$": media_mulheres,
        "Pay_Gap_%": gap_pct,
        "Estatistica_T": t_stat,
        "P_Valor": p_val,
        "Significativo": "Sim" if p_val < 0.05 else "Não"
    })

resultado_ttest = pd.DataFrame(resultados)
resultado_ttest
```

```python
resultado_ttest[
    [
        "Senioridade",
        "Media_Homens_R$",
        "Media_Mulheres_R$",
        "Pay_Gap_%",
        "Estatistica_T",
        "P_Valor",
        "Significativo"
    ]
].round({
    "Media_Homens_R$": 2,
    "Media_Mulheres_R$": 2,
    "Pay_Gap_%": 2,
    "Estatistica_T": 3,
    "P_Valor": 6
})
```
