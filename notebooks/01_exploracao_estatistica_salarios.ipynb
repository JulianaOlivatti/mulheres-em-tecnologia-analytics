# Exploração estatística dos salários

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/synthetic/fato_mercado_salarios_tech.csv")
df.head()
```

```python
df.shape
```

```python
df[["Genero", "Senioridade", "Anos_Experiencia", "Salario_Mensal_BRL"]].isna().sum()
```

```python
df["Genero"].value_counts()
```

```python
df.groupby("Genero")["Salario_Mensal_BRL"].agg(
    ["count", "mean", "median", "std", "min", "max"]
).round(2)
```

```python
df.groupby(["Senioridade", "Genero"])["Salario_Mensal_BRL"].agg(
    ["count", "mean", "median", "std"]
).round(2)
```

```python
medias = df.groupby(
    ["Senioridade", "Genero"]
)["Salario_Mensal_BRL"].mean().unstack()

medias["Gap_R$"] = medias["Masculino"] - medias["Feminino"]
medias["Gap_%"] = (medias["Gap_R$"] / medias["Masculino"]) * 100

medias.round(2)
```

```python
df.groupby("Genero")["Anos_Experiencia"].agg(
    ["count", "mean", "median", "std"]
).round(2)
```

```python
df.boxplot(column="Salario_Mensal_BRL", by="Genero")
plt.suptitle("")
plt.title("Salário mensal por gênero")
plt.xlabel("Gênero")
plt.ylabel("Salário mensal (R$)")
plt.show()
```
