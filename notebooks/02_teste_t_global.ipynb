# Teste t global

**H0:** as médias salariais de homens e mulheres são iguais.  
**H1:** as médias salariais de homens e mulheres são diferentes.

```python
import pandas as pd
from scipy import stats

df = pd.read_csv("../data/synthetic/fato_mercado_salarios_tech.csv")
```

```python
homens = df.loc[
    df["Genero"] == "Masculino",
    "Salario_Mensal_BRL"
]

mulheres = df.loc[
    df["Genero"] == "Feminino",
    "Salario_Mensal_BRL"
]

pd.DataFrame({
    "Genero": ["Masculino", "Feminino"],
    "n": [len(homens), len(mulheres)],
    "Media_R$": [homens.mean(), mulheres.mean()],
    "Mediana_R$": [homens.median(), mulheres.median()],
    "Desvio_Padrao": [homens.std(), mulheres.std()]
}).round(2)
```

```python
t_stat, p_val = stats.ttest_ind(
    homens,
    mulheres,
    equal_var=False
)

t_stat, p_val
```

```python
alpha = 0.05

resultado = "Rejeitar H0" if p_val < alpha else "Não rejeitar H0"

pd.DataFrame({
    "Estatistica_T": [t_stat],
    "P_Valor": [p_val],
    "Alpha": [alpha],
    "Resultado": [resultado]
})
```

```python
gap_absoluto = homens.mean() - mulheres.mean()
gap_percentual = (gap_absoluto / homens.mean()) * 100

pd.DataFrame({
    "Gap_R$": [gap_absoluto],
    "Gap_%": [gap_percentual]
}).round(2)
```
